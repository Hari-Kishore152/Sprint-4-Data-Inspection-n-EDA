### 5. Index

The Index is a unique label assigned to each row in a DataFrame.

By default, Pandas assigns numeric indices starting from 0.

```python
import pandas as pd

data = {
    "Name": ["Hari", "Kumar", "Ravi", "Priya"],
    "Age": [22, 25, 23, 24],
    "Salary": [25000, 35000, 30000, 40000]
}

df = pd.DataFrame(data)

print(df)
```

#### How to Inspect the Index?

```python
df.index
```

#### Assign a Custom Index

You can assign your own index labels to the rows.

```python
df.index = ["A", "B", "C", "D"]

print(df)
```

#### Reset the Index

To restore the default numerical index:

```python
df.reset_index(inplace=True)
```

| Attribute/Method            | Purpose                                                |
| --------------------------- | ------------------------------------------------------ |
| `df.index`                  | View the index                                         |
| `df.index = [...]`          | Assign custom index labels                             |
| `df.reset_index()`          | Reset the index and preserve the old index as a column |
| `df.reset_index(drop=True)` | Reset the index without preserving the old index       |
| `df.set_index("Name")`      | Set the `Name` column as the index                     |

### 8. sample()

 The sample() method in Pandas is used to select and display a random sample of rows from a DataFrame.

It is useful in Data Inspection and Exploratory Data Analysis (EDA) to examine a random subset of a dataset

```python
df.sample(n=5)
```

####  AI ML use case

In AI/ML, sample() is used to randomly select rows from a dataset to inspect data, understand its distribution, and test data quality.

####  Specify Random State

This code selects 5 random rows from a DataFrame and ensures that you can get the same selection again when you run the code with the same dataset.


```python
df.sample(5, random_state=42)
```

Understanding each part:

df → Your DataFrame (dataset).

sample(5) → Randomly selects 5 rows from the dataset.

random_state=42 → Ensures that the same random rows are selected each time you run the code, provided the dataset and sampling conditions remain unchanged.

### 10. info()

The info() method in Pandas provides a summary of a DataFrame, including the number of rows and columns, column names, non-null values, data types, and memory usage.

It helps us understand:

- Number of rows
- Number of columns
- Column names
- Data types
- Missing values
- Memory usage

#### example:


```python
import pandas as pd

data = {
    "Name": ["Hari", "Kumar", "Ravi", None],
    "Age": [22, 25, None, 24],
    "Salary": [25000, 35000, 30000, 40000]
}

df = pd.DataFrame(data)

df.info()
```

**Why is Age a float64 column?**

The Age column contains a missing value (None). Pandas commonly converts integer columns with missing values to float64 because standard NumPy integer types cannot represent missing values directly.

### 13. nunique()

The nunique() method in Pandas is used to count the number of unique (distinct) values in a Series or in each column of a DataFrame.

#### Syntax

df.nunique()

By default, it counts unique values in each column and excludes missing values (NaN).

#### example

```python
import pandas as pd

data = {
    "Name": ["Hari", "Kumar", "Ravi", "Hari", "Kumar"],
    "Age": [22, 25, 23, 22, 25],
    "City": ["Chittoor", "Bangalore", "Chittoor", "Chittoor", "Bangalore"]
}

df = pd.DataFrame(data)

print(df.nunique())
```

- Name has 3 unique values: Hari, Kumar, Ravi.
- Age has 3 unique values: 22, 25, 23.
- City has 2 unique values: Chittoor, Bangalore.

### 14. value_counts()

The value_counts() method in Pandas is used to count how many times each unique value appears in a Series or DataFrame column

#### example

```python

import pandas as pd

data = {
    "Name": ["Hari", "Kumar", "Ravi", "Hari", "Kumar", "Hari"],
    "Age": [22, 25, 23, 22, 25, 22],
    "City": ["Chittoor", "Bangalore", "Chittoor", "Chittoor", "Bangalore", "Chittoor"]
}

df = pd.DataFrame(data)

print(df["City"].value_counts())
```

```python
##Includes missing values in the counts.

df["City"].value_counts(dropna=False)
```

```python
#Sorts the counts from lowest to highest.

df["City"].value_counts(ascending=True)
```
