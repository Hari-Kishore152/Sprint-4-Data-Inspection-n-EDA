### 1. What is a Dataset?

A feature in a dataset is a characteristic or attribute that describes an observation.

**In Machine Learning, features are the input variables a model uses to make predictions.**

### 3. What is a Target?

A Target is the variable that we want to predict.

A target is the value that a Machine Learning model is trained to predict.

In a dataset, the target is usually represented by a particular column.

It is also called:


- Label
- Output Variable
- Response Variable
- Prediction Variable

- A. Numerical Target

When the target contains a numerical value, such as marks, salary, or house price, the problem is generally a regression task.

Features tell the model what information it can use; the target tells the model what value it should learn to predict.

### 4. Independent Variables
Independent Variables are the variables used to predict the target.

An independent variable is a variable that is used as an input to explain or predict changes in another variable.

Independent variables are the features (X) that we feed into an ML model to predict a target (y).

Independent variables influence or help explain the outcome, but they are not the target being predicted in that particular model.

#### Example:

Student Marks Prediction

Suppose we want to predict student marks.

- Independent variables (X): Study_Hours, Attendance
- Dependent variable (y): Marks

Here, study hours and attendance are the input variables used to predict marks.

#### Use in AI/ML

Independent variables provide the input information from which a model learns patterns.

### 5. Dependent Variables

A dependent variable is the variable whose value we want to explain or predict based on other variables.

In Machine Learning, the dependent variable is commonly called the target variable (y).

A Dependent Variable is the variable whose value depends on one or more independent variables.

#### Example:

| Prediction Task          | Independent Variables (X) | Dependent Variable (y) |
| ------------------------ | ------------------------- | ---------------------- |
| House price prediction   | House size, Location      | House price            |
| Salary prediction        | Experience, Skills        | Salary                 |
| Student marks prediction | Study hours, Attendance   | Marks                  |
| Rainfall prediction      | Temperature, Humidity     | Rainfall               |


- Dependent Variable = Target Variable = The output we want to predict.

- Independent Variables = Features = The inputs used to predict the output.

Remember: The dependent variable is called dependent because, in the analysis, its value is studied in relation to the independent variables. This does not necessarily mean that the independent variables cause the outcome.

#### Use in AI/ML

The model receives input features such as study hours and attendance.
    
During training, it learns patterns between the input features and the known marks.
    
The model uses these learned patterns to predict marks for new students.

### 6. Numerical Features

Numerical features are features in a dataset that contain numerical values representing quantities or measurements.

These values can be used for mathematical calculations and statistical analysis

#### Types of Numerical Features

Numerical features are mainly divided into two types.

**A. Discrete Features**

Discrete features contain countable values, usually whole numbers.

Examples:

- Number of students in a classroom
- Number of products sold
- Number of employees in a company
- Number of rooms in a house

| Student_Name | Number_of_Siblings |

**B. Continuous Features**

Continuous features represent measurements that can take any value within a range, subject to the precision of measurement.

Examples:

- Distance
- Time

| Student_Name | Height (cm) |

### 7. Categorical Features

Categorical features are features in a dataset that represent groups, labels, or categories rather than numerical quantities.

**They describe the type or group to which an observation belongs.**

#### Example: Student Dataset

| Student_ID | Name   | Gender | Department | Marks |
| ---------: | ------ | ------ | ---------- | ----: |
|        101 | Ravi   | Male   | CSE        |    78 |
|        102 | Priya  | Female | ECE        |    91 |
|        103 | Kumar  | Male   | CSE        |    65 |
|        104 | Anjali | Female | IT         |    84 |


In this dataset:

- Gender → Categorical feature
- Department → Categorical feature
- Name → Text identifier or descriptive field; it is not usually treated as a categorical input feature without a specific reason.
- Marks → Numerical feature

**Remember: Categorical features represent which group or category something belongs to, not how much of something there is.**

#### Types of Categorical Features

Categorical features are mainly divided into two types.

**A. Nominal Features**

Nominal features represent categories that have no natural order or ranking.

Examples:

- Gender
- Department
- Blood Group
- City
- Color

Example:

The departments CSE, ECE, and IT are nominal categories.

There is no natural ranking that makes CSE greater than ECE or ECE greater than IT.

**B. Ordinal Features**

Ordinal features represent categories that have a meaningful order or ranking.

Examples:

- Education Level
- Customer Satisfaction
- Performance Rating
- T-shirt Size

These categories have a natural order:

Poor < Average < Good < Excellent

Unlike nominal categories, ordinal categories have a meaningful ranking.

However, the difference between consecutive categories is not necessarily equal.

For example, the difference between Poor and Average might not be the same as the difference between Good and Excellent.

#### Use of Categorical Features in AI/ML

Machine learning models often require categorical features to be converted into numerical representations before they can use them.

This process is called categorical encoding.

A machine learning model may not directly understand text categories such as CSE, ECE, and IT.

We can convert these categories into numerical representations using techniques such as:

- One-Hot Encoding: Creates separate binary columns for categories.
- Ordinal Encoding: Assigns numerical codes according to the meaningful order of categories.

For nominal categories such as departments, One-Hot Encoding is often a suitable starting point.

### 10. Identifier Columns

Identifiers are mainly used to distinguish one record from another rather than to describe a meaningful characteristic of that record.

**These columns are created for identification purposes and generally do not help in prediction.**

### 13. Dataset Dimensions

Dataset dimensions refer to the number of rows and columns present in a dataset.

In simple words, dataset dimensions tell us:

- How many records are present in the dataset.
- How many columns are available in the dataset.

Formula  = Dataset Dimensions = (Rows, Columns)

For example, consider the following student dataset:

| Student_ID | Name   | Age | Department | Marks |
| ---------: | ------ | --: | ---------- | ----: |
|        101 | Ravi   |  21 | CSE        |    78 |
|        102 | Priya  |  22 | ECE        |    91 |
|        103 | Kumar  |  20 | CSE        |    65 |
|        104 | Anjali |  21 | IT         |    84 |
|        105 | Hari   |  23 | ECE        |    87 |

This dataset contains:

5 rows — representing 5 students.
    
5 columns — representing 5 attributes of each student.

Therefore, the dimensions of this dataset are:


Dimensions = (5, 5)

The first number represents the number of rows, and the second represents the number of columns.

```python
import pandas as pd

data = {
    "Student_ID": [101, 102, 103, 104, 105],
    "Name": ["Ravi", "Priya", "Kumar", "Anjali", "Rahul"],
    "Age": [21, 22, 20, 21, 23],
    "Study_Hours": [4, 6, 3, 5, 2],
    "Marks": [78, 91, 65, 84, 55]
}

df = pd.DataFrame(data)
```
