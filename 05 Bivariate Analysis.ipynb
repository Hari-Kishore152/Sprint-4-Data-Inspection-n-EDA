### What Is Bivariate Analysis?

Bivariate analysis is a statistical technique used to study the relationship between two variables.

The word bivariate can be divided into:

- Bi: Two
- Variate: Variables

Therefore, bivariate analysis means analyzing two variables together to understand whether they are related.

- We want to answer this question:

Does an employee's age have a relationship with their salary?

We can investigate this by comparing the Age and Salary columns.

If salary generally increases as age increases, the variables show a positive relationship in this dataset.

#### Why Is Bivariate Analysis Important in AI/ML?

Bivariate analysis helps us:

- Identify relationships between features.
- Understand which features might be useful for predicting a target.
- Discover differences between categories.
- Identify unusual patterns and outliers.
- Support feature selection during machine learning.

Example: If we want to predict house prices, we might study the relationship between house area and price.

#### Types of Bivariate Analysis

We classify bivariate analysis according to the types of variables being compared.

| Type                       | First Variable | Second Variable | Example                       |
| -------------------------- | -------------- | --------------- | ----------------------------- |
| Numerical vs Numerical     | Numerical      | Numerical       | Age vs Salary                 |
| Numerical vs Categorical   | Numerical      | Categorical     | Department vs Salary          |
| Categorical vs Categorical | Categorical    | Categorical     | Department vs Employment Type |


#### example:


```python
# Import required libraries 
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

| Library    | Purpose                           |
| ---------- | --------------------------------- |
| Pandas     | Create and analyze datasets       |
| NumPy      | Generate numerical data           |
| Matplotlib | Create visualizations             |
| Seaborn    | Create statistical visualizations |


```python
data = {
    "Employee_ID": range(101, 121),

    "Age": [
        22, 24, 25, 27, 28,
        30, 32, 35, 26, 29,
        31, 33, 24, 28, 36,
        27, 34, 25, 30, 32
    ],

    "Experience": [
        0, 1, 2, 3, 4,
        5, 7, 10, 2, 4,
        6, 8, 1, 3, 11,
        3, 9, 2, 5, 7
    ],

    "Salary": [
        20000, 23000, 25000, 28000, 30000,
        35000, 42000, 60000, 26000, 32000,
        38000, 48000, 22000, 29000, 65000,
        27000, 55000, 24000, 36000, 45000
    ],

    "Department": [
        "IT", "HR", "IT", "Sales", "Finance",
        "IT", "Finance", "IT", "HR", "Sales",
        "Finance", "IT", "HR", "Sales", "IT",
        "Finance", "IT", "HR", "Sales", "Finance"
    ],

    "Employment_Type": [
        "Full-time", "Part-time", "Full-time", "Full-time",
        "Contract", "Full-time", "Full-time", "Full-time",
        "Part-time", "Contract", "Full-time", "Full-time",
        "Part-time", "Full-time", "Full-time", "Contract",
        "Full-time", "Part-time", "Full-time", "Full-time"
    ]
}

df = pd.DataFrame(data)

display(df)
```

#### 1.Numerical vs Numerical

Numerical vs Numerical analysis studies the relationship between two numerical variables.

Examples:

- Age vs Salary
- Experience vs Salary
- Income vs Spending

#### Age vs Salary

We want to investigate:

Do older employees generally earn higher salaries?

#### - Create a scatter plot

A scatter plot represents each observation as a point on a graph.

A scatter plot shows the relationship between two numerical variables.

- X-axis: Age
- Y-axis: Salary
- Each point: One employee

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Age",
    y="Salary"
)

plt.title("Age vs Salary")
plt.xlabel("Age (Years)")
plt.ylabel("Salary (₹)")
plt.tight_layout()
plt.show()
```

#### How to interpret the scatter plot

Look at the overall direction of the points.

Positive relationship: The points generally move upward from left to right. Higher values of age tend to occur alongside higher salaries.

Negative relationship: The points generally move downward from left to right. Higher values of one variable tend to occur alongside lower values of the other.

No clear relationship: The points do not show a consistent upward or downward pattern.

- For our employee dataset, age and salary show a generally positive pattern, although the relationship is not perfect.

#### 2.Numerical vs Categorical

This analysis compares a numerical variable across different categories.

Examples:

- Department vs Salary
- Education Level vs Salary
- Product Category vs Sales
- City vs House Price

In our dataset:

Salary is numerical.
    
Department is categorical.

We want to answer:

Do employees in different departments have different salary distributions?

#### First, calculate the average salary for each department.

```python
#To display the result in a sorted DataFrame:

department_salary = (
    df.groupby("Department", as_index=False)["Salary"]
      .mean()
      .sort_values("Salary", ascending=False)
)

department_salary
```

#### How to interpret the result

The output shows the average salary of employees in each department.

- A higher average means that department has a higher average salary in this sample.
- A lower average means that department has a lower average salary in this sample.

However, the average alone does not show how salaries vary among employees.

For example, two departments could have the same average salary but very different salary distributions.

Let's visualize the distributions.

#### Lets create a Box Plot for department and salary


A box plot summarizes the distribution of a numerical variable across categories.

It displays:

- The median.
- The first quartile, or Q1.
- The third quartile, or Q3.
- The interquartile range, or IQR.
- Whiskers, which usually extend to the most extreme observations within 1.5 IQR of the quartiles.
- Potential outliers, displayed as individual points under the usual settings.

The interquartile range is  :  IQR=Q3−Q1

```python
plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df,
    x="Department",
    y="Salary"
)

plt.title("Salary Distribution by Department")
plt.xlabel("Department")
plt.ylabel("Salary (₹)")
plt.tight_layout()
plt.show()
```

#### How to interpret the box plot

**Compare the medians:** A higher median indicates that the typical salary, measured by the median, is higher in that department.

**Compare the boxes:** A taller box indicates a larger spread across the middle 50% of salaries.

**Look at the whiskers and individual points:** These help reveal the spread and potential outliers.

For example, if the IT department has a higher median salary than HR, we can say that the median salary for IT employees is higher in this sample.

We should not conclude that every IT employee earns more than every HR employee.

#### AI/ML application

Box plots can help analysts identify:

- Salary differences across departments.
- Unusually high or low transaction amounts.
- Differences in model prediction errors across customer groups.
- Potential outliers before training a model.

#### lets create a violin plot for department and salary:

A violin plot shows the distribution of a numerical variable across categories using a shape based on the estimated data density.

In simple terms, it helps us understand where values are concentrated.

- Wider regions represent values with greater estimated density.
- Narrower regions represent values with lower estimated density.

Unlike a box plot, a violin plot provides a more detailed view of the distribution's shape.

```python
plt.figure(figsize=(9, 5))

sns.violinplot(
    data=df,
    x="Department",
    y="Salary",
    inner="quartile",
    cut=0
)

plt.title("Salary Distribution by Department")
plt.xlabel("Department")
plt.ylabel("Salary (₹)")
plt.tight_layout()
plt.show()
```

#### How to interpret the violin plot

Compare the shapes of the violins.

For example:

- A violin concentrated at higher salary values suggests that more observations are concentrated around those values.
- A wide region at a particular salary range indicates higher estimated density in that region.
- A long shape may suggest a wider distribution.

Because this dataset contains only 20 employees, the estimated density shapes should be interpreted cautiously. Violin plots are generally more informative with larger datasets.

| Feature                           | Box Plot                  | Violin Plot                 |
| --------------------------------- | ------------------------- | --------------------------- |
| Shows median                      | Yes                       | With `inner="quartile"`     |
| Shows quartiles                   | Yes                       | With `inner="quartile"`     |
| Shows distribution shape          | Limited                   | Yes                         |
| Helps identify potential outliers | Yes                       | Not its primary purpose     |
| Best use                          | Compare spread and median | Compare distribution shapes |


#### 3. Categorical vs Categorical

Categorical vs Categorical analysis investigates the relationship between two categorical variables.

Examples:

- Department vs Employment Type
- Gender vs Product Preference
- Education Level vs Job Role
- Product Category vs Customer Segment

Let's analyze:

Is the distribution of employment types different across departments?

#### lets create the **Cross tabulation** for employment type and the department

Cross tabulation, also called a contingency table, summarizes the frequency of observations for combinations of two categorical variables.

For example, it can show how many full-time, part-time, and contract employees belong to each department.

```python
pd.crosstab(
    df["Department"],
    df["Employment_Type"]
)
```

- Each row represents a department, and each column represents an employment type.

- Each cell contains the number of employees belonging to that combination.

How to interpret the output

Suppose the IT row has:

- A high count under Full-time.
- A low count under Part-time.
- No count under Contract.

This would indicate that most IT employees in the sample are full-time.

```python
#To see row percentages instead of counts:

pd.crosstab(
    df["Department"],
    df["Employment_Type"],
    normalize="index"
).mul(100).round(2)
```

#### **Grouped Bar chart** for Employment type and department 

```python
A grouped bar chart compares values across categories using bars placed next to one another.

For our example:

X-axis: Department
Bar groups: Employment Type
Y-axis: Number of employees
```

```python
plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    x="Department",
    hue="Employment_Type"
)

plt.title("Employment Type by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.legend(title="Employment Type")
plt.tight_layout()
plt.show()
```

#### How to interpret the chart

Compare the bars within each department.

For example:

- If the Full-time bar is tallest in IT, full-time employees are the most common employment type in IT.
- If HR has more part-time employees than another department, HR has a larger part-time count in this sample.

Remember that counts and percentages answer different questions. Counts show how many employees belong to each group; percentages show the proportion of each department represented by that group.



#### AI/ML application

Cross tabulation and grouped bar charts can help us investigate:

- Customer segment vs product preference.
- Customer type vs purchase decision.
- Region vs subscription status.
- Department vs employee category.

These methods are especially useful when exploring categorical features before machine learning.
